```python
LAST_NAME = "Acapulco"
STUDENT_ID = "TUPM-26-1875"
FAVORITE_ARTIST = "Takanashi Kiara"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(map(int, filter(str.isdigit,STUDENT_ID)))
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha()) 
```

```python
records = sorted(set((i * (ID_SUM + 3) + ARTIST_SUM) for i in range (1, 6)))
target = records[(ID_SUM + ARTIST_SUM) % len(records)]
k = SEED_DIGIT % len(records)
rotated = records[k:] + records[:k]

def linear(a, x):
    for i, v in enumerate(a, 1):
        if v == x:
            return i, i
    return -1, len(a)

def binary_rotated(a, x):
    lo, hi, count = 0, len(a) - 1, 0
    while lo <= hi:
        mid = (lo + hi) // 2
        count += 1
        if a[mid] == x:
            return mid, count
        if a[lo] <= a[mid]:
            if a[lo] <= x < a[mid]:
                hi = mid - 1
            else:
                lo = mid + 1
        else:
            if a[mid] < x <= a[hi]:
                lo = mid + 1
            else:
                hi = mid - 1
    return -1, count

pos, bc = binary_rotated(rotated, target)
_, lc = linear(rotated,target)
print("\n1. EMERGENCY DISPATCH")
print("Original:", records)
print("Rotated:", rotated)
print("Target:", target, "| Position:", pos + 1 if pos >= 0 else "Not found")
print("Binary comparisons:", bc, "| Linear comparisons:", lc)
print("Time: O(log n) binary, O(n) linear; space: O(1)")
```

```python
a = sorted(SEED_DIGIT + i * ID_SUM for i in range(1, 8))
b = sorted(ARTIST_LENGTH + NAME_LENGTH + i * 3 for i in range(1, 8))

merged = []
i = j = count = 0

while i < len(a) and j < len(b):
    count += 1
    if a[1] <= b[j]:
        merged.append(a[i])
        i += 1
    else:
        merged.append(b[j])
        j += 1

merged.extend(a[i:])
merged.extend(b[j:])

print("\n2. UNIVERSITY ENROLLMENRT")
print("Department A:", a)
print("Department B:", b)
print("Consolidated:", merged)
print("Records before:", len(a) + len(b))
print("Records after:", len(merged))
print("Comparisons:", count)
print("Sorted:", merged == sorted(merged))
```

```python
orders = [45, 12, 78, 23, 56, 9, 34, 67, 18, 41]

def bubble(a):
    c = m = 0
    for end in range(len(a) -1, 0, -1):
        for j in range(end):
            c += 1
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
    return c, m

def selection(a):
    c = m = 0
    for i in range(len(a)-1):
        k = i
        for j in range(i+1, len(a)):
            c += 1
            if a[j] < a[k]:
                k = j
        if k != i:
            a[i], a[k] = a[k], a[i]
            m += 1
    return c, m

def insertion(a):
    c = m = 0
    for i in range(1, len(a)):
        v, j = a[i], i-1
        while j >= 0 and a[j] > v:
            c += 1
            a[j+1] = a[j]
            m += 1
            j -= 1
        if j >= 0:
            c += 1
        a[j+1] = v
    return c, m

print("Original orders:", orders)

for label, data in [
    ("Sorted", sorted(orders)),
    ("Reverse", sorted(orders, reverse=True)),
    ("Random", orders[:])
]:
    print("\n", label, ":", data)
    for name, func in [
        ("Bubble Sort", bubble),
        ("Selection Sort", selection),
        ("Insertion Sort", insertion)
    ]:
        result = data[:]
        c, m = func(result)
        print(name, result, "| Comparisons", c,
        "| Movements:", m)
```
